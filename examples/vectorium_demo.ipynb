# Anima‑Vectorium Demo Notebook 📘
This notebook demonstrates the expressive → geometric pipeline:
- expressive state creation (💠)
- persistence lineage (🗄️)
- read‑vault inspection (🔒)
- expressive → geometric projection (🪢)
- kernel execution (🧩)
- diagnostics + safety (🔍 ⚖️ 🛡️)


```python
from anima.anima_core import AnimaCore
from anima.persistence import Persistence
from anima.read_vault import ReadVault
from bridge.bridge_readonly import ReadOnlyBridge
from bridge.bridge_kernel import KernelBridge
from vectorium.kernel import Kernel

print("Initialized Anima‑Vectorium demo environment.")
```

## 1. Create expressive states (💠)

```python
anima = AnimaCore()
persistence = Persistence()
vault = ReadVault(persistence)

s1 = anima.create("hello‑vectorium")
s2 = anima.create({"msg": "expressive‑geometry"})

s1, s2
```

## 2. Store expressive states (🗄️)

```python
r1 = persistence.store(s1)
r2 = persistence.store(s2)

print(f"Stored expressive states: {r1.index}, {r2.index}")
```

## 3. Inspect read‑vault (🔒)

```python
vault.list()
```

## 4. Project expressive → geometric (🪢)

```python
readonly_bridge = ReadOnlyBridge(anima, vault)
packet = readonly_bridge.project(0)
packet
```

## 5. Kernel execution (🧩)

```python
kernel = Kernel()
kernel_bridge = KernelBridge(kernel)

result = kernel_bridge.execute(packet, a={"x": 1}, b={"x": 2})
result
```

## 6. Diagnostics + Safety (🔍 ⚖️ 🛡️)

```python
print("Diagnostics:", result.diagnostics)
print("Safety:", result.safety)
```
